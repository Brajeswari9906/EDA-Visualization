```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv('Titanic.csv')
```


Question 1 : Create a countplot to show the number of passengers by gender.



Answer:
Countplot is used to visualize the number of male and female passengers.


```python

plt.figure(figsize=(6,4))

sns.countplot(x='Sex', data=df)

plt.title('Number of Passengers by Gender')

plt.xlabel('Gender')

plt.ylabel('Count')

plt.show()
```


Interpretation:
The chart shows the distribution of male and female passengers on the Titanic.



Question 2 : Create a bar chart to show the number of passengers in each passenger class using different colors.



Answer:
Bar chart helps compare passenger counts across different classes.


```python

class_counts = df['Pclass'].value_counts().sort_index()

plt.figure(figsize=(6,4))

plt.bar(
    class_counts.index,
    class_counts.values,
    color=['blue', 'green', 'orange']
)

plt.title('Passengers in Each Class')

plt.xlabel('Passenger Class')

plt.ylabel('Number of Passengers')

plt.show()
```


Interpretation:
Passenger distribution varies across first, second, and third class.



Question 3 : Create a histogram to show the distribution of passenger age.



Answer:
Histogram helps understand the distribution of passenger ages.


```python

plt.figure(figsize=(7,4))

plt.hist(df['Age'], bins=20)

plt.title('Distribution of Passenger Age')

plt.xlabel('Age')

plt.ylabel('Frequency')

plt.show()
```


Interpretation:
Most passengers belong to the young and middle-age groups.



Question 4 : Create a boxplot to detect outliers in the fare column.



Answer:
Boxplot helps identify fare outliers.


```python

plt.figure(figsize=(7,4))

sns.boxplot(x=df['Fare'])

plt.title('Boxplot of Fare')

plt.xlabel('Fare')

plt.show()
```


Interpretation:
Extreme fare values indicate outliers and expensive ticket prices.



Question 5 : Create a scatterplot to

a. show the relationship between age and fare.

b. analyze the relationship between family size (SibSp + Parch) and fare.



Answer:
Scatterplots help analyze relationships between variables.


```python

plt.figure(figsize=(7,4))

plt.scatter(df['Age'], df['Fare'])

plt.title('Age vs Fare')

plt.xlabel('Age')

plt.ylabel('Fare')

plt.show()
```

```python

df['Family_Size'] = df['SibSp'] + df['Parch']

plt.figure(figsize=(7,4))

plt.scatter(df['Family_Size'], df['Fare'])

plt.title('Family Size vs Fare')

plt.xlabel('Family Size')

plt.ylabel('Fare')

plt.show()
```


Interpretation:
Fare generally increases for some passengers with larger family sizes and certain age groups.



Question 6 : Create a countplot to show survival count by gender.



Answer:
Countplot compares survival counts based on gender.


```python

plt.figure(figsize=(7,4))

sns.countplot(x='Sex', hue='Survived', data=df)

plt.title('Survival Count by Gender')

plt.xlabel('Gender')

plt.ylabel('Count')

plt.show()
```


Interpretation:
Female passengers generally had a higher survival rate compared to male passengers.



Question 7 : Create a bar chart to show survival by passenger class.



Answer:
Bar chart helps analyze survival patterns across passenger classes.


```python

survival_class = df.groupby('Pclass')['Survived'].mean()

plt.figure(figsize=(6,4))

plt.bar(
    survival_class.index,
    survival_class.values,
    color=['purple', 'green', 'orange']
)

plt.title('Survival by Passenger Class')

plt.xlabel('Passenger Class')

plt.ylabel('Average Survival')

plt.show()
```


Interpretation:
Passengers in higher classes had better survival chances.



Question 8 : Create a heatmap to show correlation between numerical variables.



Answer:
Heatmap helps visualize relationships between numerical features.


```python

plt.figure(figsize=(10,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```


Interpretation:
Strong positive and negative relationships can be identified between variables.



Question 9 : Identify which passenger class paid the highest fare on average.



Answer:
Average fare is calculated for each passenger class.


```python

average_fare = df.groupby('Pclass')['Fare'].mean()

print(average_fare)

highest_class = average_fare.idxmax()

print("\nPassenger Class with Highest Average Fare:", highest_class)
```


Interpretation:
First-class passengers generally paid the highest average fare.
